## Introduction

The goal of this notebook is to **provide access to Yahoo API, downlaod and save price data**.  

## Import librairies

```python
import pandas as pd
import yfinance as yf
```

## Inputs

```python
ticker = "SPY"
start_date = "1999-01-01"
end_date = "2025-12-01"
```

## Script

```python
data = yf.download(ticker, start=start_date, end=end_date, progress=False, auto_adjust=False)['Adj Close']
data.index.name = None
data.columns.name = None
data.columns = ['price']
data.to_csv(f"data/{ticker}.csv")
```
