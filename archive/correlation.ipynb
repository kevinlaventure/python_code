```python
import numpy as np
import pandas as pd
import plotly.express as px
```

```python
symbols = ['SPY', 'EWQ']
```

```python
df = pd.DataFrame()
for symbol in symbols:
    df_temp = pd.read_csv(f'data/{symbol}.csv', index_col=0, parse_dates=True)['price']
    df = pd.concat([df, df_temp], axis=1)
df.columns = symbols
```

```python
df = df.dropna()
```

```python
x = df.rolling(window=5).apply(lambda x: x[-1]/x[0]-1 )
```

```python
x = x.dropna()
```

```python
x['spread'] = x['SPY']-x['EWQ']
```

```python
x['beta'] = x['EWQ']/x['SPY']
```

```python
x.sort_values('SPY').head(20)*100
```

```python
px.scatter(x.set_index(symbols[0])['spread'])
```
